# Python Funksiyalar Tapşırıqları


## Tapşırıq 1 - Salam funksiyası

`salam()` adlı funksiya yarat.

Funksiya ekrana:

`Salam, xoş gəlmisən!`

göstərsin.

Funksiyanı 3 dəfə çağır.

## Tapşırıq 2 - İstifadəçi Adı

`helloUser(name)` adlı funksiya yarat.

Funksiya istifadəçinin adını qəbul terminalda göstərsin:

`Salam, Murad!`

Funksiyanı fərqli adlarla ən azı 3 dəfə çağır.

## Tapşırıq 3 - Hesab

`hesab(icki_qiymet, sirniyyat_qiymet)` adlı funksiya yarat.

Funksiya ümumi məbləği hesablayıb terminalda göstərsin.

## Task 4 - Return-dən istifadə

`topla(a, b)` adlı funksiya yarat.

Funksiya iki ədədi toplayıb `return` etsin.

Sonra nəticəni ayrıca `print()` ilə göstər.

## Task 5 - Film

`filmIzle(movie='Spiderman')` adlı funksiya yarat.

Əgər parametr göndərilməsə default olaraq `Interstellar` istifadə olunsun.

Ekrana:

`Bu gün izlənəcək film: Interstellar`

yazdırsın.

## Tapşırıq 6 - Online Mağaza

`hesabla(mehsul, qiymet, say)` adlı funksiya yarat.

Funksiya məhsulun ümumi qiymətini hesablayıb qaytarsın (`return`).

Məsələn:

Telefon - 2 ədəd - 4000 AZN

## Tapşırıq 7 - Temperatur Çevirici

`tempraturCevir(celsius)` adlı funksiya yarat.

Funksiya temperaturu Fahrenheitə çevirib return etsin.

Formula:

`(C * 9/5) + 32`

## Task 8 - Oyun

`xarakterSec(ad, silah='Qılınc')` adlı funksiya yarat.

Funksiya belə yazdırsın:

`Araz adlı oyunçu Qılınc istifadə edir.`

## Tapşırıq 9 - Maaş

`maasHesabla(umumi_saat, saat_basi_gelir)` adlı funksiya yarat.

İş saatına görə maaşı hesablayıb return etsin.

Əgər saatlıq ödəniş verilməsə default olaraq 15 istifadə olunsun.
