## Tapşırıq 10 — Tələbə Qiyməti hesablamaq

3. Sinifdəki bütün qiymətlərin **ortalamasını** hesabla və sonda cavabı göstər 
4. Ən yüksək qiyməti alan tələbənin adını terminalda göstər

```python
qiymətlər = [92, 78, 88, 55, 95, 63, 81]
```

**Gözlənilən nəticə:**
```
1. Əli     — 92 ✅ Keçdi

4. Orxan   — 55 ❌ Kəsildi
5. Günel   — 95 ✅ Keçdi
6. Tural   — 63 ❌ Kəsildi
7. Sevinc  — 81 ❌ Kəsildi

Sinif ortalama qiyməti: 78.86
Ən yüksək qiymət: Günel — 95
```

````python
telebeler = ["Əli", "Leyla", "Nigar", "Orxan", "Günel", "Tural", "Sevinc"]
qiymətlər = [92, 78, 88, 55, 95, 63, 81]
#  Kodunu bura yaz
````
