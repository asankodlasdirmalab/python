## Tapşırıq 9 — BMİ Hesablayıcı

**Şərt:** İstifadəçidən **boy** (sm) və **çəki** (kq) götür
BMİ (Bədən Kütlə İndeksi) formulunu tətbiq et: `bmi = çəki / (boy_metr ** 2)`  
Nəticəyə görə kateqoriyanı göstər:

- BMİ < 18.5 → **Arıq**
- 18.5 ≤ BMİ < 25 → **Normal**
- 25 ≤ BMİ < 30 → **Artıq çəkili**
- BMİ ≥ 30 → **Obez**

**Nümunə:**
```

## Tapşırıq 10 — Bank

**Şərt:** Başlanğıc balans `1000` AZN-dir.  
İstifadəçidən əməliyyat növünü daxil etməsini istə: `"artir"` və ya `"gotur"`  
Sonra məbləği daxil etməsini istə.

Aşağıdakı şərtləri yoxla:
- Əgər əməliyyat **`"artir"`** dırsa:
  - Məbləğ 0-dan böyük olmalıdır → balansı artır və yeni balansı ekranda göstər
  - Deyilsə → `"Məbləğ sıfırdan böyük olmalıdır!"`
- Əgər əməliyyat **`"gotur"`** dürsə:
  - Məbləğ balansdan az və ya bərabər olmalıdır → balansdan çıxar və yeni balansı ekranda göstər
  - Deyilsə → `"Balansınız kifayət deyil!"` yazılsın
- Başqa bir əməliyyat daxil edilərsə → `"Yanlış əməliyyat!"` yazılsın

```
Əməliyyat (artir / gotur): gotur
