## Tapşırıq 9 —  Məktəb Qiymət Sistemi

Aşağıda bütün şagirdlərin dərs məlumatları verilib  

Data forması: `(ad, (riyaziyyat, fizika, ingilis, tarix))` şəklindədir

```python
telebeler = (
    ('Aytən',  (90, 85, 78, 92)),
    ('Rəşad',  (55, 48, 70, 60)),
    ('Ləman',  (100, 95, 88, 97)),
    ('Tural',  (40, 35, 50, 45)),
    ('Nigar',  (75, 80, 65, 70)),
)
```

Proqram aşağıdakıları etsin:
1. Hər şagirdin **orta qiymətini** hesabla
2. Orta **60-dan aşağı** olanları **«Zəif»**, **60–84** arası **«Kafi»**, **85+** **«Əla»** kimi göstər
3. «Əla» qiymət alan tələbələri ayrıca bir **set**-də saxla və terminalda **```Əlaçı```** şagirdlər kimi göstər
4. Ən yüksək ortalamaya sahib şagird göstər və ən ağıllı şagird kimi göstər

````python
telebeler = (
    ('Aytən',  (90, 85, 78, 92)),
    ('Rəşad',  (55, 48, 70, 60)),
    ('Ləman',  (100, 95, 88, 97)),
    ('Tural',  (40, 35, 50, 45)),
    ('Nigar',  (75, 80, 65, 70)),
)
# Kodu burda yaz
````
